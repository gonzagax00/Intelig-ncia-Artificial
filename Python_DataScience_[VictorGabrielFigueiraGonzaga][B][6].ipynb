**1-Exercícios de NumPy**

```python
# 1. Crie um array unidimensional de 50 elementos contendo números inteiros aleatórios entre 10 e 100.

import numpy as np

array = np.random.randint(10, 101, 50)
print(array)
```

```python
# 2. Gere uma matriz 4x4 com números inteiros de 1 a 16 e extraia a última linha.

import numpy as np

matriz = np.arange(1,17).reshape(4,4)
print(matriz[-1])
```

```python
# 3. Crie um array de zeros com comprimento 15 e substitua os valores da 5ª à 10ª posição por 1.

import numpy as np

array = np.zeros(15)
array[4:10] = 1
print(array)
```

```python
# 4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.

import numpy as np

array = np.random.randint(1, 50, 100)

mediana = np.median(array)

valores, contagens = np.unique(array, return_counts=True)
moda = valores[np.argmax(contagens)]

print("Mediana:", mediana)
print("Moda:", moda)
```

```python
# 5. Crie um array de 20 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 3 por -3.

import numpy as np

array = np.random.randint(1, 101, 20)

array[array % 3 == 0] = -3

print(array)
```

```python
# 6. Gere uma matriz 5x5 de números aleatórios entre 0 e 1 e encontre a soma de cada coluna.

import numpy as np

matriz = np.random.rand(5, 5)

soma_colunas = np.sum(matriz, axis=0)

print("Matriz 5x5: \n", matriz)
print("Soma de cada coluna: \n", soma_colunas)
```

```python
# 7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.

import numpy as np

array1 = np.random.rand(40)
array2 = np.random.rand(40)

correlacao = np.corrcoef(array1, array2)[0, 1]

print(correlacao)
```

```python
# 8. Gere uma matriz 3x3 com números inteiros aleatórios e eleve todos os seus elementos ao quadrado.

import numpy as np

matriz = np.random.randint(1, 10, (3, 3))
print("Matriz original: \n", matriz)

matriz_quadrado = matriz ** 2
print("Matriz elevada ao quadrado: \n", matriz_quadrado)

```

```python
# 9. Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.

import numpy as np

array = np.random.rand(10)
print("Array original: \n", array)

array_normalizado = (array - np.min(array)) / (np.max(array) - np.min(array))
print("Array normalizado: \n", array_normalizado)
```

```python
# 10. Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.

import numpy as np

matriz = np.eye(6)
print("Matriz identidade: \n", matriz)
matriz[np.diag_indices_from(matriz)]
print("Diagonal principal: \n", matriz)
```

```python
# 11. Gere um array de 100 números inteiros e encontre o valor que mais se repete no array.

import numpy as np

array = np.random.randint(1, 10, 100)
print("Array: \n", array)
valores, contagens = np.unique(array, return_counts=True)
valor_mais_repetido = valores[np.argmax(contagens)]
print("Valor mais repetido: \n", valor_mais_repetido)
```

```python
# 12. Crie uma matriz 4x4 e substitua os valores da diagonal principal por 7.

import numpy as np

matriz = np.random.randint(1, 10, (4, 4))
print("Matriz: \n", matriz)
np.fill_diagonal(matriz, 7)
print("Matriz com diagonal principal 7: \n", matriz)
```

```python
# 13. Crie um array contendo os 10 primeiros números pares e calcule sua soma cumulativa.

import numpy as np

array = np.arange(2, 22, 2)
print("Array: \n", array)
soma_cumulativa = np.cumsum(array)
print("Soma cumulativa: \n", soma_cumulativa)

```

```python
# 14. Calcule a diferença entre dois arrays de números aleatórios de mesmo comprimento e encontre a média das diferenças.

import numpy as np

array = np.random.randint(1, 10, 10)
array2 = np.random.randint(1, 10, 10)
print("Array: \n", array)
print("Array2: \n", array2)
diferenca = np.abs(array - array2)
print("Diferença: \n", diferenca)
```

```python
# 15. Gere um array contendo 20 números e substitua todos os valores maiores que 50 por 50.

import numpy as np

array = np.random.randint(1, 100, 20)
print("Array: \n", array)
array[array > 50] = 50
print("Array com valores maiores que 50: \n", array)
```

```python
# 16. Crie um array unidimensional e reorganize seus valores em uma matriz 3x4.

import numpy as np

array = np.random.randint(1, 100, 12)
print("Array: \n", array)
matriz = array.reshape(3, 4)
print("Matriz: \n", matriz)
```

```python
# 17. Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada valor por sua posição na matriz.

import numpy as np

matriz = np.random.randint(1, 10, (4, 4))
print("Matriz: \n", matriz)
matriz_multiplicada = matriz * np.indices(matriz.shape)[0]
print("Matriz multiplicada: \n", matriz_multiplicada)
```

```python
# 18. Crie um array de 10 elementos contendo números entre 1 e 100 e substitua todos os valores múltiplos de 5 por 0.

import numpy as np

array = np.random.randint(1, 101, 10)
print("Array: \n", array)
array[array % 5 == 0] = 0
print("Array com valores múltiplos de 5: \n", array)
```

```python
# 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ãngulo entre eles (produto escalar normalizado).

import numpy as np

array = np.random.randint(1, 10, 20)
array2 = np.random.randint(1, 10, 20)
print("Array: \n", array)
print("Array2: \n", array2)
produto_escalar = np.dot(array, array2)
print("Produto escalar: \n", produto_escalar)
```

```python
# 20. Crie uma matriz de números inteiros e encontre o valor mínimo de cada linha.

import numpy as np

matriz = np.random.randint(1, 10, (4, 4))
print("Matriz: \n", matriz)
valor_minimo = np.min(matriz, axis=1)
print("Valor mínimo de cada linha: \n", valor_minimo)
```
