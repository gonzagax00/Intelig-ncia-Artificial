```python
# 3-Solicite o raio de um círculo e calcule sua área. Use a fórmula área = π * r²

# Biblioteca usada para trazer o valor de π.
import math

# Pede para usuário o raio e já converte para número decimal.
# Usei o float pois ele ja representa casas decimais.
# Se usasse int não contaria a informação depois do ponto.
raio = float(input("Digite o raio do círculo: "))

# Calcula a área usando: área = π * r².
# Usando math.pi você acessa a biblioteca importada e puxa o valor de pi.
area = math.pi * (raio ** 2)

# Mostra o resultado do calculo.
print("A área do círculo é:", area)
```

```python
# 5-Crie um programa que imprima a tabuada de um número fornecido pelo usuário.

# Pede ao usuario o número da tabuada desejada.
n = int(input("Digite um número que deseja para ver a tabuada: "))

# Como é uma formula que vai se repetir varias vezes,
# para não ficar um código muito longo, decidi usar o FOR
# para criar um loop e reduzir as linhas de código.
# A cada volta do loop, a variavel que chamei de i vira outro valor,
# até chegar no valor que queremos.
# Vale lembrar que em python o valor para uma casa antes do valor digitado,
# assim explicando o número 11 no fim do FOR.
for i in range(1, 11):

  # Aqui guaramos o valor da multiplicação da tabuada na variavel resultado.
  resultado = n * i

  # Ele vai mostrar o resultado usando o f para trocar o valor
  # dentro das chaves para o valor real que representa cada variavel.
  print(f"{n} X {i} = {resultado}")

# Escolhi usar o FOR porque se fosse fazer "manual" o código ficaria grande,
# e sabendo que podemos simplificar e otimizar o programa não faz sentido
# criar varias linhas com contas desnecessárias.
```

```python
# 6-Faça um programa que exiba todos os números pares de 1 a 20.

# Usei novamente o FOR para fazer a contagem de 1 até 20.
# Poderia fazer manual, ja sabendo quais são os pares,
# porem ficaria uma lista grande de código, que não é o que queremos.
for n in range(1,21):

  # Dentro do if usamos o calculo para ver se o resto da divisão é igual a 0.
  # O % retorna o resto da divisão para o if e ele verifica se é true ou false.
  # Se for True, ele executa o valor e se for False ele pula.
  if n % 2 == 0:
    print(n,"é um numero par")
```

```python
# 7-Solicite um número ao usuário e informe se ele é par ou ímpar.

# Solicita ao usuário um numero inteiro.
n = int(input("Digite um número: "))

# Usei if para comparar se o número digitado pelo usuário é par ou ímpar.
# Se o resto da divisão do número for 0 então ele printa que é par.
if n % 2 == 0:
  print(n,"é um numero par")

# Se o resto da divisão for diferente de 0 então ele printa ímpar.
else:
  print(n,"é um numero impar")
```

```python
# 8-Crie um programa que verifique se uma palavra é um palíndromo (ex: "radar").

# Aqui peço para o usuário digitar uma palavra.
palavra = input("Digite uma palavra: ")

# Aqui tem uma coisa interessante, se eu uso esse programa sem o lower,
# mesmo se a palavra for palíndromo porem com espaços e acentos,
# ele da erro e fala que não é palíndromo, o lower converte todo o
# texto para minuscúlo e sem espaço antes de comparar.
palavra_invertida = palavra.lower()

# Nessa parte ele faz a comparação se a palavra é palíndromo ou não.
# Usei o slicing [::-1] para dividir a string e pegar todos os caracteres,
# porem ele faz isso de trás para frente com o -1 do final.
# Caso a palavra seja a mesma invertida, ele printa que é uma palavra palíndromo.
if palavra_invertida == palavra_invertida[::-1]:
  print(palavra,"é um palíndromo")
# Caso não seja, ele printa que não é um palíndromo.
else:
  print(palavra,"não é um palíndromo")


```

```python
# 9-Calcule o fatorial de um número fornecido pelo usuário.

# Importo a biblioteca math que ja tem a função pronta de calcular fatorial.
import math

# Peço para o usuário digitar um número para iniciar o calculo.
n = int(input("Digite um número para calcular o vetorial: "))

# Aqui a função math.fatorial ja tem um loop que faz a conta para nós,
# sem que precisemos fazer o FOR manualmente.
# Ele recebe o número que o usuário digitou e devolve o fatorial pronto.
# que fica armazenada nele e apenas printamos depois.
fatorial = math.factorial(n)
print(f"O fatorial de {n} é {fatorial}")
```

```python
# 12-Conte o número de ocorrências de cada caractere em uma string fornecida.

# Peço para o usuário digitar uma pavara ou frase.
texto = input("Digite uma palavra ou frase: ")

# Criei um dicionario pois ele guarda os dados digitado em pares,
# uma chave e um valor, no caso, cada letra seria uma chave, e o
# número de letras seria o valor.
contador = {}

# Nessa parte a "letra" verifica todo o texto digitado pelo usuário.
# O if e else verifica se a letra ja existe no dicionário, se sim, ele soma mais 1 no contador,
# se não, ele cria uma chave nova no dicionário.
for letra in texto:
  if letra in contador:
    contador[letra] = contador[letra] + 1
  else:
    contador[letra] = 1

# Printei apenas para mostrar como fica no dicionário.
print(contador)

# Aqui faço um print mais elaborado, o FOR divide o dicionário em letra e quantidade,
# usando o .items(), e exibe na ordem (letra, quantidade).
for letra, quantidade in contador.items():
  print(f"A letra '{letra}' aparece {quantidade} vezes.")

```

```python
# 15-Calcule a média dos números em uma lista.

# Utilizei o sum pois ele vai fazer a soma de todos os números da lista,
# e o len pois ele me retorna quantos elementos tem dentro da lista.
# Criei a variavel "media" pra fazer o calculo dos números da lista.
num = [10, 34, 89, 23, 90, 54, 65, 49, 73]

soma = sum(num)
quantidade = len(num)
media = soma / quantidade

print("A média dos números é: ", media)
```

```python
# 17-Conte a frequência de cada palavra em uma string fornecida pelo usuário.

texto = input("Digite uma frase: ")

# Usei o .split(), pois ele divide a string em palavras utilizando o espaço e não caracteres.
palavras = texto.split()

# Usei a mesma lógica do exercício 12 de contar caracteres, porem agora com palavras.
frequencia = {}

# No FOR ele vai analisar palavra por palavra ao invés da frase toda.
# adicionando a frequencia da palavra no dicionario, como por exemplo:
# se a palavra ja existe, ela soma mais 1, se não, cria uma chave com o valor 1.
for palavra in palavras:
  if palavra in frequencia:
    frequencia[palavra] = frequencia[palavra] + 1
  else:
    frequencia[palavra] = 1

for palavra, quantidade in frequencia.items():
  print(f"A palavra '{palavra}' aparece {quantidade} vezes.")
```

```python
# 19-Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.

import random

# O import random gera um número aleatório entre 1 e 100,
# que foi colocado no random.randint(1, 100).
num_aleatorio = random.randint(1, 100)

# Comecei com 0 só pra garantir que o while entra pelo menos uma vez,
# já que ele checa a condição antes de rodar o bloco.
tentativa = 0

# usei o WHILE aqui porque não dá pra saber quantas tentativas o usuário vai precisar até acertar.
# Esse loop fica repetindo enquanto a tentativa for diferente do número escolhido do usuário.
while tentativa != num_aleatorio:
  tentativa = int(input("Digite um número entre 1 e 100: "))

  if tentativa == num_aleatorio:
    print("Parabéns, você acertou!")
  elif tentativa < num_aleatorio:
    print("Tente um número maior.")
  else:
    print("Tente um número menor.")

```

```python
# 20-Dada uma matriz (lista de listas), crie sua matriz transposta.

# matriz original, cada lista de dentro é uma linha
matriz = [
    [90, 48, 82],
    [75, 55, 37]
]

# Criei essa lista para guardar o resultado final da transposta.
transposta = []

# pra cada coluna eu percorro as linhas pegando o elemento dela,
# junto tudo numa lista nova e jogo isso dentro da transposta
for coluna in range(len(matriz[0])):
    nova_linha = []
    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])
    transposta.append(nova_linha)

print("\nMatriz transposta:")
for linha in transposta:
    print(linha)
```
