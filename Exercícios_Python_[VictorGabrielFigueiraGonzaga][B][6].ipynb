```python
# 21-Implemente a busca binária para encontrar um elemento em uma lista ordenada.

# Só funciona em lista ordenada. a cada passo, olho o elemento do
# meio e descarto metade da lista, em vez de percorrer item por item.

def busca_binaria(lista, alvo):
  inicio = 0
  fim = len(lista) - 1

  while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == alvo:
      return meio
    elif lista[meio] < alvo:
      inicio = meio + 1
    else:
      fim = meio - 1

  return -1

lista_ordenada = [1, 3, 5, 7, 9, 11, 13, 17, 19]
alvo = int(input("Digite o número que quer buscar: "))
resultado = busca_binaria(lista_ordenada, alvo)

if resultado != -1:
  print(f"Número encontrado no índice {resultado}")
else:
  print("Número não encontrado na lista")
```

```python
# 22-Crie uma função para criptografar uma mensagem usando a cifra de César.

# Desloca cada letra um número fixo de posições no alfabeto (ex:
# deslocamento 3 faz 'a' virar 'd'). Uso % 26 pra "dar a volta"
# quando passa de 'z'.

def cifrar_cesar(texto, deslocamento):
  resultado = ""
  for letra in texto:
    if letra.isalpha():
      base = ord('A') if letra.isupper() else ord('a')
      nova_letra = chr((ord(letra) - base + deslocamento) % 26 + base)
      resultado += nova_letra
    else:
      resultado += letra
  return resultado

mensagem = input("Digite a mensagem para cifrar: ")
deslocamento = int(input("Digite o deslocamento: "))
mensagem_cifrada = cifrar_cesar(mensagem, deslocamento)

print(f"Mensagem cifrada: {mensagem_cifrada}")
```

```python
# 23-Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

# Mesma lógica da cifra, só que aplicando o deslocamento contrário
# pra desfazer o embaralhamento das letras.

def decifrar_cesar(texto, deslocamento):
  resultado = ""
  for letra in texto:
    if letra.isalpha():
      base = ord('A') if letra.isupper() else ord('a')
      nova_letra = chr((ord(letra) - base - deslocamento) % 26 + base)
      resultado += nova_letra
    else:
      resultado += letra
  return resultado

mensagem_cifrada = input("Digite a mensagem cifrada: ")
deslocamento = int(input("Digite o deslocamento usado na cifra: "))
mensagem_original = decifrar_cesar(mensagem_cifrada, deslocamento)

print(f"Mensagem original: {mensagem_original}")
```

```python
# 24-Implemente uma função para validar um CPF brasileiro.

# CPF válido segue uma regra nos dois últimos dígitos.
# Calculo eles com base nos 9 primeiros
# números e comparo com o que foi digitado.

def validar_cpf(cpf):
  cpf = cpf.replace(".", "").replace("-", "")

  if len(cpf) != 11 or cpf == cpf[0] * 11:
    return False

  soma = 0
  for i in range(9):
    soma += int(cpf[i]) * (10 - i)
  resto = soma % 11
  digito1 = 0 if resto < 2 else 11 - resto

  if digito1 != int(cpf[9]):
    return False

  soma = 0
  for i in range(10):
    soma += int(cpf[i]) * (11 - i)
  resto = soma % 11
  digito2 = 0 if resto < 2 else 11 - resto

  if digito2 != int(cpf[10]):
    return False

  return True

cpf_digitado = input("Digite o CPF (com ou sem pontuação): ")

if validar_cpf(cpf_digitado):
  print("CPF válido")
else:
  print("CPF inválido")
```

```python
# 30-Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

# Sorteia uma palavra da lista e vai revelando as letras certas que
# o usuário digitar. o jogo acaba quando ele descobre a palavra
# inteira ou quando as tentativas acabam

import random

palavras = ["python", "programacao", "faculdade", "computador"]
palavra_secreta = random.choice(palavras)
letras_descobertas = ["_"] * len(palavra_secreta)
tentativas = 6

print("Jogo da Forca!")

while tentativas > 0 and "_" in letras_descobertas:
  print(" ".join(letras_descobertas))
  letra = input("Digite uma letra: ").lower()

  if letra in palavra_secreta:
    for i in range(len(palavra_secreta)):
      if palavra_secreta[i] == letra:
        letras_descobertas[i] = letra
  else:
    tentativas -= 1
    print(f"Letra errada! Tentativas restantes: {tentativas}")

if "_" not in letras_descobertas:
  print(f"Parabéns! Você descobriu a palavra: {palavra_secreta}")
else:
  print(f"Você perdeu! A palavra era: {palavra_secreta}")
```

```python
# 31-Implemente o algoritmo de ordenação por bolha (Bubble Sort).

# Ele compara elementos vizinhos e troca de posição quando estão fora
# de ordem, repetindo isso várias vezes até tudo ficar ordenado,
# cada passada o maior valor vai até o final da lista

numeros = [23, 4, 42, 8, 16, 15]
n = len(numeros)

for i in range(n):
  for j in range(n - i - 1):
    if numeros[j] > numeros[j + 1]:
      numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(f"Lista ordenada: {numeros}")
```

```python
# 32-Crie um jogo da velha para dois jogadores.

# Faço uma checagem se alguma combinação de linha/coluna/diagonal
# está completa com o mesmo símbolo.

tabuleiro = [" " for _ in range(9)]

def mostrar_tabuleiro():
  print(f"{tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
  print("--+---+--")
  print(f"{tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
  print("--+---+--")
  print(f"{tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")

def verificar_vencedor():
  combinacoes = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),
    (0, 3, 6), (1, 4, 7), (2, 5, 8),
    (0, 4, 8), (2, 4, 6)
  ]
  for a, b, c in combinacoes:
    if tabuleiro[a] == tabuleiro[b] == tabuleiro[c] != " ":
      return tabuleiro[a]
  return None

jogador_atual = "X"
jogadas = 0

while jogadas < 9:
  mostrar_tabuleiro()
  posicao = int(input(f"Jogador {jogador_atual}, escolha uma posição (0-8): "))

  if tabuleiro[posicao] == " ":
    tabuleiro[posicao] = jogador_atual
    jogadas += 1

    vencedor = verificar_vencedor()
    if vencedor:
      mostrar_tabuleiro()
      print(f"Jogador {vencedor} venceu!")
      break

    jogador_atual = "O" if jogador_atual == "X" else "X"
  else:
    print("Posição já ocupada, escolha outra")

if jogadas == 9 and not verificar_vencedor():
  mostrar_tabuleiro()
  print("Empate!")
```

```python
# 33-Verifique se duas palavras fornecidas são anagramas uma da outra.

# Em vez de ordenar as letras como no exercício 27, monto um
# dicionário contando quantas vezes cada letra aparece em cada
# palavra, e comparo se os dois dicionários são iguais.

palavra1 = input("Digite a primeira palavra: ").lower().replace(" ", "")
palavra2 = input("Digite a segunda palavra: ").lower().replace(" ", "")

if len(palavra1) != len(palavra2):
  print("As palavras não são anagramas")
else:
  contador1 = {}
  contador2 = {}

  for letra in palavra1:
    contador1[letra] = contador1.get(letra, 0) + 1

  for letra in palavra2:
    contador2[letra] = contador2.get(letra, 0) + 1

  if contador1 == contador2:
    print("As palavras são anagramas")
  else:
    print("As palavras não são anagramas")
```

```python
# 35-Gere a sequência de Fibonacci até um número n fornecido pelo usuário.

# Aqui o critério de parada é o valor do número (parar quando passar
# de n), não uma quantidade fixa de termos - por isso uso while.

n = int(input("Digite um número limite: "))
a, b = 0, 1

while a <= n:
  print(a, end=" ")
  a, b = b, a + b
```

```python
# 36-Crie um gerador de senhas aleatórias com letras e números.

# Monta uma senha sorteando caracteres aleatórios de uma string que
# junta letras maiúsculas, minúsculas e números.

import random
import string

tamanho = int(input("Digite o tamanho da senha: "))
caracteres = string.ascii_letters + string.digits
senha = ""

for i in range(tamanho):
  senha += random.choice(caracteres)

print(f"Senha gerada: {senha}")
```

```python
# 37-Implemente uma função recursiva para calcular o fatorial de um número.

# Recursão é quando a função chama ela mesma. Sempre precisa de um
# "caso base" (aqui, n ser 0 ou 1) pra parar, senão ela chamaria a
# si mesma pra sempre e travaria o programa

def fatorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * fatorial(n - 1)

numero = int(input("Digite um número: "))
print(f"O fatorial de {numero} é {fatorial(numero)}")
```

```python
# 40-Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

# Menu com while True que fica repetindo até o usuário escolher
# sair (break), permitindo adicionar, remover e listar itens.

lista_compras = []

while True:
  print("\n1 - Adicionar item")
  print("2 - Remover item")
  print("3 - Listar itens")
  print("4 - Sair")
  opcao = input("Escolha uma opção: ")

  if opcao == "1":
    item = input("Digite o item que quer adicionar: ")
    lista_compras.append(item)
  elif opcao == "2":
    item = input("Digite o item que quer remover: ")
    if item in lista_compras:
      lista_compras.remove(item)
    else:
      print("Item não encontrado na lista")
  elif opcao == "3":
    print("Lista de compras:", lista_compras)
  elif opcao == "4":
    break
  else:
    print("Opção inválida")
```

```python
# 42-Gere todos os números primos até um número n fornecido.

# Reaproveita a lógica de verificar se um número é primo (testando
# se ele divide por algo entre 2 e ele mesmo), usando isso pra
# cada número de 2 até n e guardando os primos numa lista

n = int(input("Digite um número limite: "))
primos = []

for numero in range(2, n + 1):
  eh_primo = True
  for i in range(2, numero):
    if numero % i == 0:
      eh_primo = False
      break
  if eh_primo:
    primos.append(numero)

print(f"Números primos até {n}: {primos}")
```

```python
# 46-Calcule a média e o desvio padrão de uma lista de números.

# Desvio padrão mede o quanto os números se espalham em torno da
# média. A fórmula é a raiz quadrada da média das diferenças ao
# quadrado entre cada número e a média.

import math

numeros = [10, 34, 89, 23, 90, 54, 65, 49, 73]

media = sum(numeros) / len(numeros)

soma_diferencas = 0
for numero in numeros:
  soma_diferencas += (numero - media) ** 2

desvio_padrao = math.sqrt(soma_diferencas / len(numeros))

print(f"Média: {media:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")
```

```python
# 47-Conte quantas vezes cada número aparece em uma lista.

# Mesma lógica de contar letras/palavras que já vimos, só que
# usando números como chave do dicionário em vez de texto

numeros = [9, 4, 4, 3, 7, 7, 7, 5]
contagem = {}

for numero in numeros:
  if numero in contagem:
    contagem[numero] += 1
  else:
    contagem[numero] = 1

for numero, quantidade in contagem.items():
  print(f"O número {numero} aparece {quantidade} vez(es)")
```

```python
# 48-Implemente o algoritmo de ordenação por inserção (Insertion Sort).

# Pega cada número da lista e "encaixa" ele na posição certa entre
# os que já foram ordenados antes dele, empurrando os maiores uma
# posição pra frente até achar o lugar certo.

numeros = [23, 4, 42, 8, 16, 15]

for i in range(1, len(numeros)):
  atual = numeros[i]
  j = i - 1
  while j >= 0 and numeros[j] > atual:
    numeros[j + 1] = numeros[j]
    j -= 1
  numeros[j + 1] = atual

print(f"Lista ordenada: {numeros}")
```

```python
# 49-Calcule a soma dos dígitos de um número inteiro fornecido.

# Uso % 10 pra pegar sempre o último dígito do número, e // 10
# (divisão inteira) pra "cortar" esse dígito e seguir com o resto.

numero = int(input("Digite um número: "))
soma = 0
numero_temp = numero

while numero_temp > 0:
  digito = numero_temp % 10
  soma += digito
  numero_temp //= 10

print(f"A soma dos dígitos de {numero} é {soma}")
```

```python
# 50-Calcule a frequência de cada letra em uma string fornecida pelo usuário.

# Uso isalpha() pra contar só letras (ignorando espaço e pontuação).

texto = input("Digite uma palavra ou frase: ").lower()
frequencia = {}

for letra in texto:
  if letra.isalpha():
    if letra in frequencia:
      frequencia[letra] += 1
    else:
      frequencia[letra] = 1

for letra, quantidade in frequencia.items():
  print(f"A letra '{letra}' aparece {quantidade} vez(es)")
```
