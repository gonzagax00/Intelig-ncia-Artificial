**2 - Exercícios de SciPy**

```python
# 1. Calcule a integral da função f(x) = x 2 sin(x) no intervalo de 0 a π.

import numpy as np
from scipy import integrate

def f(x):
    return x**2 * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)
print("Resultado da integral:", resultado)
print('Erro estimado: ', erro)

```

```python
# 2. Resolva a equação diferencial dy dx = x − y, com y(0) = 1, no intervalo de 0 a 10.

from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(x, y):
    return x - y

solucao = solve_ivp(eq_diferencial, [0, 10], [1], t_eval=np.linspace(0, 10, 100))
print(f'Tempos: {solucao.t}')
print(f'Solução y: {solucao.y[0]}')
```

```python
# 3. Encontre as raízes da função f(x) = x 4 − 3x 3 + 2 utilizando SciPy.

from scipy.optimize import root

def f(x):
    return x**4 - 3*x**3 + 2

resultado = root(f, x0=-1)
print("Raiz:", resultado.x[0])
```

```python
# 4. Use a função de otimização de SciPy para maximizar f(x) = −x 2+ 4x+ 1.

from scipy.optimize import minimize

def f(x):
    return -x**2 + 4*x + 1

resultado = minimize(lambda x: -f(x), x0=0)
print("x que maximiza f(x):", resultado.x[0])
print("Valor máximo de f(x):", f(resultado.x[0]))
```

```python
# 5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

from scipy.fft import fft
import numpy as np

t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

frequencias = fft(sinal)
print("Magnitude dos 5 primeiros componentes:", np.abs(frequencias[:5]))
```

```python
# 6. Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.

from scipy.linalg import inv
import numpy as np

matriz_5x5 = np.random.rand(5, 5)
inversa = inv(matriz_5x5)

print(f'Matriz:\n{matriz_5x5}')
print(f'Matriz Inversa:\n{inversa}')
print(f'Verificação (matriz × inversa):\n{np.dot(matriz_5x5, inversa)}')
```

```python
# 7. Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

from scipy.interpolate import interp1d
import numpy as np

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

f_cubica = interp1d(x, y, kind='cubic')
x_novos = np.array([0.5, 1.5, 2.5, 3.5])
y_interpolados = f_cubica(x_novos)
print("Interpolados: ", y_interpolados)
```

```python
# 8. Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6.

from scipy import linalg
import numpy as np

A = np.array([[3, 2], [1, 4]])
B = np.array([5, 6])

solucao = linalg.solve(A, B)
print("Solução [x, y]:", solucao)
```

```python
# 9. Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).

from scipy.optimize import curve_fit
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

def parabola(x, a, b, c):
    return a*x**2 + b*x + c

parametros, _ = curve_fit(parabola, x, y)
print("Parâmetros [a, b, c]:", parametros)
```

```python
# 19. Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica.

import pandas as pd

dados = {
    'Produto': ['Mouse', 'Teclado', 'Monitor', 'Webcam', 'Headset'],
    'Preço': [50.0, 120.0, 800.0, 150.0, 200.0],
    'Quantidade': [10, 5, 3, 8, 6]
}
df = pd.DataFrame(dados)

media_colunas = df.mean(numeric_only=True)
print(media_colunas.to_string())
```

```python
# 20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma coluna numérica.

import pandas as pd

dados = {
    'Produto': ['Mouse', 'Teclado', 'Monitor', 'Cadeira', 'Mesa'],
    'Categoria': ['Informática', 'Informática', 'Informática', 'Móveis', 'Móveis'],
    'Preço': [50.0, 120.0, 800.0, 450.0, 600.0]
}
df = pd.DataFrame(dados)

mediana_por_categoria = df.groupby('Categoria')['Preço'].median()
print(mediana_por_categoria.to_string())
```

```python
# 6. Plote dois gráficos de linha em subplots diferentes: y1 = e x e y2 = log(x).

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0.1, 5, 100)
y1 = np.exp(x)
y2 = np.log(x)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,4))
ax1.plot(x, y1, color='#e63946')
ax1.set_title('y = e^x')
ax2.plot(x, y2, color='#2a9d8f')
ax2.set_title('y = log(x)')
plt.tight_layout()
plt.show()
```

```python
# 8. Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
erro = 0.1

plt.plot(x, y, label='y = sin(x)')
plt.fill_between(x, y-erro, y+erro, alpha=0.3, label='Intervalo de confiança')
plt.legend()
plt.show()
```

```python
# 10. Plote três gráficos de linha em um único gráfico, cada um com uma cor e estilo de linha diferentes.

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1, y2, y3 = np.sin(x), np.cos(x), np.sin(x)*np.cos(x)

plt.plot(x, y1, linestyle='-', label='sin(x)')
plt.plot(x, y2, linestyle='--', label='cos(x)')
plt.plot(x, y3, linestyle=':', label='sin(x)*cos(x)')
plt.legend()
plt.show()
```
